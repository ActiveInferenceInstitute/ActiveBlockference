# Gridworld walkthrough

This notebook demonstrates the current ActiveBlockference API with deterministic local data and no external services. It uses explicit multi-agent starts, a seed, backend parity, and a trajectory validation report.

```python
from blockference import ExperimentConfig, GridWorld
from blockference.io import validate_trajectory_dataframe
from blockference.simulations import run_experiment

world = GridWorld(3, {0: (0, 0)})
print(world.step({0: 3}))
config_data = {"name": "notebook", "seed": 7, "grid": {"planning_length": 1}, "simulation": {"timesteps": 1, "runs": 1, "n_agents": 2, "target": [2, 2], "initial_states": [[0, 0], [2, 2]]}, "output": {"path": None}}
config = ExperimentConfig.from_dict({**config_data, "engine": "radcad"})
radcad = run_experiment(config)
cadcad = run_experiment(ExperimentConfig.from_dict({**config_data, "engine": "cadcad"}))
assert radcad["env_states"].astype(str).tolist() == cadcad["env_states"].astype(str).tolist()
assert validate_trajectory_dataframe(radcad).ok
print(config.grid.dimension, len(radcad), "backend parity and validation passed")
```
